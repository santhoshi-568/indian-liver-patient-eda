```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("indian_liver_patient.csv")

print("Dataset loaded successfully!")
```

```python
# Display first five rows
df.head()
```

```python
# Dataset shape
print("Rows and Columns:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
df.info()
```

```python
# Check missing values
print(df.isnull().sum())

# Statistical summary
df.describe()# Check missing values
print(df.isnull().sum())

# Statistical summary
df.describe()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Age',
    bins=20,
    kde=True
)

plt.title('Distribution of Patient Age')
plt.xlabel('Age')
plt.ylabel('Number of Patients')

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='Gender',
    hue='Gender'
)

plt.title('Gender Distribution of Patients')
plt.xlabel('Gender')
plt.ylabel('Number of Patients')

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    data=df,
    x='Dataset',
    y='Total_Bilirubin',
    hue='Dataset'
)

plt.title('Total Bilirubin Levels by Disease Status')
plt.xlabel('Patient Category')
plt.ylabel('Total Bilirubin')

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='Age',
    y='Total_Bilirubin',
    hue='Dataset',
    alpha=0.7
)

plt.title('Age vs Total Bilirubin')
plt.xlabel('Age')
plt.ylabel('Total Bilirubin')

plt.show()
```
